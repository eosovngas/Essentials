# 1. Títulos

Los títulos se crean utilizando el símbolo `#`.

```markdown
# Título 1
## Título 2
### Título 3
#### Título 4
##### Título 5
###### Título 6
```

Mientras más símbolos `#`, menor será el nivel del título.

# 13. Ejemplo aplicado a un proceso ETL

Una buena práctica es utilizar Markdown para explicar **qué hace cada etapa del notebook y por qué existe**.

## Objetivo

Realizar la carga y transformación de información de clientes desde archivos Parquet hacia una tabla Delta.

## Flujo

1. Leer archivos de origen.
2. Validar estructura.
3. Limpiar columnas.
4. Aplicar reglas de negocio.
5. Guardar información en formato Delta.

## Entrada

| Parámetro | Valor |
|---|---|
| Formato | Parquet |
| Origen | Data Lake |
| Entidad | Clientes |

## Reglas de negocio

- El identificador del cliente no puede ser nulo.
- El nombre debe almacenarse en mayúsculas.
- La fecha de alta debe tener formato `yyyy-MM-dd`.
- Los registros duplicados deben eliminarse.

## Salida

La información será almacenada en la tabla `TB_CLIENTES`.

`````python
# Ejemplo de código que puede acompañar la documentación Markdown
from pyspark.sql import functions as F

# Lectura del origen
df = spark.read.parquet('/ruta/origen/clientes')

# Transformación de ejemplo
df = df.withColumn('NOMBRE', F.upper(F.col('NOMBRE')))

display(df)
`````

# 15. Buenas prácticas

### ✅ Recomendado

- Utilizar títulos para separar claramente las etapas.
- Documentar el objetivo antes del código.
- Explicar las reglas de negocio.
- Documentar entradas y salidas.
- Utilizar tablas para parámetros y estructuras.
- Mantener una estructura consistente entre notebooks.
- Utilizar bloques de código para mostrar ejemplos sin ejecutarlos.

### ❌ Evitar

- Crear bloques enormes de texto.
- Documentar obviedades que no aportan valor.
- Repetir comentarios que ya están explicados en el código.
- Utilizar demasiados emojis o estilos diferentes.
- Mezclar documentación técnica y funcional sin estructura.
