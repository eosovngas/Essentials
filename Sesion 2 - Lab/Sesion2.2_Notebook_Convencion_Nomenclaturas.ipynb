# Convención de Nomenclaturas

Notebook para estandarizar y generar abreviaciones utilizadas en objetos de datos y Microsoft Fabric.

## Reglas
1. **Frases con varias palabras:** se toma la primera letra de cada palabra.
   - `Eduardo Osorio` → `EO`
   - `Microsoft Fabric` → `MF`
   - `Azure Data Factory` → `ADF`

2. **Palabras individuales:** se divide la palabra en dos segmentos y se toma la primera letra de cada segmento.
   - `Notebook` → `NB`
   - `Pipeline` → `PL`
   - `Lakehouse` → `LH`
   - `Warehouse` → `WH`

> Nota 1: la segunda regla es una convención personalizada y debe mantenerse consistente en todos los proyectos.

> Nota 2: Para nombres compuestos, utilizar las iniciales de cada palabra. Para palabras simples, dividir la palabra en dos segmentos de longitud equivalente y utilizar la primera letra de cada segmento.

````python
def abreviar(nombre):
    palabras = nombre.split()

    if len(palabras) > 1:
        resultado = ""

        for palabra in palabras:
            resultado += palabra[0]

        return resultado.upper()

    palabra = palabras[0]
    mitad = len(palabra) // 2

    return (palabra[0] + palabra[mitad]).upper()
````

````python
ejemplos = [
    'Eduardo Osorio',
    'Microsoft Fabric',
    'Azure Data Factory',
    'Notebook',
    'Pipeline',
    'Lakehouse',
    'Warehouse',
    'Dataflow',
    'Semantic Model'
]

for ejemplo in ejemplos:
    print(f'{ejemplo:25} -> {abreviar(ejemplo)}')
````

## Catálogo recomendado para Microsoft Fabric

| Objeto | Abreviación | Ejemplo |
|---|---|---|
| Notebook | `NB` | `NB_CargaClientes` |
| Pipeline | `PL` | `PL_CargaClientes` |
| Dataflow | `DF` | `DF_TransformacionClientes` |
| Lakehouse | `LH` | `LH_Comercial` |
| Warehouse | `WH` | `WH_Finanzas` |
| Tabla | `TB` | `TB_CLIENTES` |
| Vista | `VW` | `VW_VENTAS_MENSUALES` |
| Stored Procedure | `SP` | `SP_CARGA_VENTAS` |
| Semantic Model | `SM` | `SM_VENTAS` |

## Estructura sugerida

```text
<TIPO>_<DOMINIO>_<PROCESO>_<DETALLE>
```

Ejemplos:

```text
NB_COMERCIAL_CARGA_CLIENTES
PL_COMERCIAL_CARGA_VENTAS
LH_COMERCIAL
TB_COMERCIAL_CLIENTES
VW_COMERCIAL_VENTAS_MENSUALES
```
