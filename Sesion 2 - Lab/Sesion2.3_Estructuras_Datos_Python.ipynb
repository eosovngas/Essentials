# 🐍 Estructuras de Datos en Python

## Listas, diccionarios y estructuras anidadas

**Objetivo:** comprender cómo almacenar, consultar, modificar y recorrer información en Python.

> 💡 Idea clave: **una lista organiza elementos por posición; un diccionario organiza información mediante claves.**

## 1. ¿Qué es una estructura de datos?

Una estructura de datos es una forma de **organizar y almacenar información** para poder trabajar con ella.

En esta clase veremos principalmente:

- `list` → colección ordenada de elementos.
- `dict` → información organizada como `clave: valor`.
- Lista de diccionarios → estructura muy utilizada para representar registros.

## 2. Listas (`list`)

Una lista permite almacenar varios elementos en un orden determinado.

```python
frutas = ["manzana", "pera", "naranja"]
```

### ⚠️ Importante
Python comienza a contar las posiciones desde **0**.

````python
frutas = ["manzana", "pera", "naranja", "plátano"]

print(frutas)
````

### Acceder a elementos

Usamos el índice del elemento:

`lista[0]` → primer elemento  
`lista[1]` → segundo elemento  
`lista[2]` → tercer elemento

````python
print("Primer elemento:", frutas[0])
print("Tercer elemento:", frutas[2])
````

### `len()` y `append()`

- `len()` permite conocer cuántos elementos tiene una lista.
- `append()` agrega un elemento al final.

````python
print("Cantidad:", len(frutas))

frutas.append("uva")

print("Después de agregar:", frutas)
````

### Modificar elementos

Las listas son **mutables**: podemos cambiar sus elementos.

````python
frutas[1] = "kiwi"

print(frutas)
````

### Recorrer una lista con `for`

````python
for fruta in frutas:
    print("Fruta:", fruta)
````

## 3. Diccionarios (`dict`)

Un diccionario almacena información mediante pares:

**clave → valor**

Ejemplo:

```python
persona = {
    "nombre": "Ana",
    "edad": 30
}
```

````python
persona = {
    "nombre": "Ana",
    "edad": 30,
    "profesion": "Ingeniera"
}

print(persona)
````

### Acceder a valores

A diferencia de una lista, utilizamos la **clave**.

````python
print("Nombre:", persona["nombre"])
print("Edad:", persona["edad"])
````

### Modificar y agregar información

````python
persona["edad"] = 31
persona["ciudad"] = "Santiago"

print(persona)
````

### Recorrer un diccionario

Con `.items()` podemos obtener simultáneamente la clave y el valor.

````python
for clave, valor in persona.items():
    print(clave, "→", valor)
````

## 4. Lista de diccionarios ⭐

Esta estructura aparece constantemente cuando trabajamos con **APIs, JSON, archivos y datos**.

Tenemos una lista y cada elemento de la lista es un diccionario.

````python
alumnos = [
    {"nombre": "Juan", "edad": 20, "nota": 6.2},
    {"nombre": "María", "edad": 21, "nota": 5.8},
    {"nombre": "Pedro", "edad": 19, "nota": 6.5}
]

print(alumnos)
````

### ¿Cómo accedemos a un dato?

```python
alumnos[0]["nombre"]
```

Se lee como:

1. `alumnos[0]` → primer elemento de la lista.
2. `["nombre"]` → valor de la clave `nombre`.

````python
print("Primer alumno:", alumnos[0])
print("Nombre:", alumnos[0]["nombre"])
print("Nota:", alumnos[0]["nota"])
````

### Recorrer todos los alumnos

````python
for alumno in alumnos:
    print(alumno["nombre"], "→", alumno["nota"])
````

## 5. Aplicar lógica: filtrar información

Podemos combinar estructuras de datos con condiciones `if`.

````python
for alumno in alumnos:
    if alumno["nota"] >= 6.0:
        print(alumno["nombre"], "aprobó con", alumno["nota"])
````

## 6. Ejemplo aplicado: productos

Ahora usemos una estructura similar a la que podríamos encontrar en un sistema empresarial.

````python
productos = [
    {"nombre": "Notebook", "precio": 800000, "stock": 5},
    {"nombre": "Mouse", "precio": 25000, "stock": 10},
    {"nombre": "Teclado", "precio": 45000, "stock": 0}
]

for producto in productos:
    print(producto["nombre"], "→ $", producto["precio"])
````

## 7. Ejercicio guiado 🧑‍💻

Crea una lista llamada `clientes` con 3 diccionarios.

Cada cliente debe tener:

- `nombre`
- `edad`
- `ciudad`

### Desafíos

1. Imprimir el nombre del segundo cliente.
2. Cambiar la ciudad del primer cliente.
3. Agregar un cuarto cliente.
4. Recorrer todos los clientes.
5. Mostrar solamente clientes mayores de 30 años.

````python
# Intenta resolver el ejercicio antes de mirar la solución.

clientes = [
    {"nombre": "Carlos", "edad": 28, "ciudad": "Santiago"},
    {"nombre": "Laura", "edad": 35, "ciudad": "Valparaíso"},
    {"nombre": "Pedro", "edad": 42, "ciudad": "Concepción"}
]

# 1. Segundo cliente
print(clientes[1]["nombre"])

# 2. Modificar ciudad del primer cliente
clientes[0]["ciudad"] = "Maipú"

# 3. Agregar cuarto cliente
clientes.append({
    "nombre": "Ana",
    "edad": 31,
    "ciudad": "Santiago"
})

# 4 y 5. Recorrer y filtrar
for cliente in clientes:
    if cliente["edad"] > 30:
        print(cliente["nombre"], "→", cliente["edad"], "años")
````

## 8. Resumen 🎯

| Estructura | ¿Cómo organiza los datos? | Ejemplo de acceso |
|---|---|---|
| Lista | Por posición | `frutas[0]` |
| Diccionario | Por clave | `persona["nombre"]` |
| Lista + diccionario | Posición + clave | `alumnos[0]["nombre"]` |

### Para recordar

**Lista → posición**  
**Diccionario → clave**  
**Lista de diccionarios → registros**
