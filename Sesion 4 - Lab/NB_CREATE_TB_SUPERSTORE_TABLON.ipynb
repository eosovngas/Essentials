```python
import pandas as pd
```

```python
path_file_pd = "/lakehouse/default/Files/EDUARDO OSORIO/XLSX/SuperStore_Tablon.xlsx"
path_file_sp = "Files/EDUARDO OSORIO/XLSX/SuperStore_Tablon.xlsx"
table_name = "superstore_tablon"
```

# Leer tablas con pandas y crear con pyspark

```python
# Leer Excel con Pandas
df_pd = pd.read_excel(
    path_file_pd,
    engine="openpyxl"
)

print(f"Registros: {len(df_pd)}")
print(f"Columnas: {len(df_pd.columns)}")

display(df_pd.head())
```

# 

# Convertir Pandas → PySpark

```python
df_sp = spark.createDataFrame(df_pd)

display(df_sp)
```

# Crear la tabla Delta

```python
(
    df_sp.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable(table_name)
)

print(f"Tabla '{table_name}' creada correctamente.")
```
