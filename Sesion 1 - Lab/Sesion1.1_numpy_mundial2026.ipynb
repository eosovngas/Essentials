# 🔢 Introducción a NumPy
## Con los datos de Portugal y Argentina — Mundial 2026 🏆

Este notebook enseña los conceptos fundamentales de **NumPy** usando estadísticas  
de los jugadores de Portugal y Argentina en el FIFA World Cup 2026.

## 0. Importar NumPy

```python
import numpy as np
print(f"Versión de NumPy: {np.__version__}")
```

---
## 1. Datos de los planteles como listas Python

Extraemos atributos numéricos de cada jugador. Luego los convertimos en **arrays de NumPy**.


```python
# ── Portugal ─────────────────────────────────────────────────
nombres_por = [
    "Diogo Costa","Rui Silva","José Sá",
    "Rúben Dias","Gonçalo Inácio","João Cancelo","Nuno Mendes","Diogo Dalot","Nélson Semedo","Tomás Araújo","Renato Veiga",
    "Bruno Fernandes","Vitinha","João Neves","Rúben Neves","Bernardo Silva","Matheus Nunes","Samú Costa",
    "Cristiano Ronaldo","Gonçalo Ramos","Rafael Leão","Pedro Neto","João Félix","Gonçalo Guedes","Francisco Conceição","Rafa Silva"
]
edades_por  = [26,30,31, 27,24,31,23,26,31,23,22, 30,25,21,28,30,26,23, 41,24,26,25,26,28,23,31]
caps_por    = [42, 9, 7, 75,22,68,35,38,42,14,10, 81,40,20,56,92,30, 8, 217,28,44,38,50,58,18,47]
goles_por   = [ 0, 0, 0,  5, 1, 4, 1, 3, 1, 0, 0, 16, 3, 2, 4,18, 1, 0, 136,13, 9, 7,11,10, 3, 8]

# ── Argentina ─────────────────────────────────────────────────
nombres_arg = [
    "Emiliano Martínez","Gerónimo Rulli","Juan Musso",
    "Cristian Romero","Lisandro Martínez","Nicolás Otamendi","Nahuel Molina","Nicolás Tagliafico","Gonzalo Montiel","Valentín Barco",
    "Lionel Messi","Alexis Mac Allister","Rodrigo De Paul","Enzo Fernández","Leandro Paredes","Giovani Lo Celso","Exequiel Palacios","Nico Paz","Thiago Almada",
    "Lautaro Martínez","Julián Álvarez","Ángel Di María","Paulo Dybala","Alejandro Garnacho","Nicolás González","Luka Romero"
]
edades_arg  = [32,32,31, 27,27,38,27,32,28,20, 38,26,31,24,31,29,26,21,24, 27,25,38,33,21,27,21]
caps_arg    = [58,14,17, 50,38,119,49,65,39, 7, 193,48,82,44,77,58,32,12,14, 72,52,142,55,18,35, 6]
goles_arg   = [ 0, 0, 0,  4, 3,  8, 6, 4, 2, 0, 112,12,13, 5, 7, 8, 3, 2, 2,  33,25, 31,36, 7,10, 1]

print(f"Portugal: {len(nombres_por)} jugadores | Argentina: {len(nombres_arg)} jugadores")
```

---
## 2. Crear arrays de NumPy — `np.array()`

Un **array** de NumPy es como una lista Python, pero mucho más rápido para cálculos  
matemáticos y vectorizados.


```python
# Convertir listas a arrays NumPy
edades_por_arr  = np.array(edades_por)
caps_por_arr    = np.array(caps_por)
goles_por_arr   = np.array(goles_por)

edades_arg_arr  = np.array(edades_arg)
caps_arg_arr    = np.array(caps_arg)
goles_arg_arr   = np.array(goles_arg)

print(f"Tipo:  {type(edades_por_arr)}")
print(f"dtype: {edades_por_arr.dtype}")
print(f"shape: {edades_por_arr.shape}")
print(f"Array: {edades_por_arr}")
```

---
## 3. Arrays 2D (matrices)

Podemos apilar los datos como filas para crear una **matriz**.  
Cada fila = un atributo, cada columna = un jugador.


```python
# Matriz de estadísticas de Portugal: 3 atributos × 26 jugadores
stats_por = np.array([edades_por, caps_por, goles_por])
print(f"Shape de stats_por: {stats_por.shape}  → (atributos × jugadores)")
print()
print("── Primeros 6 jugadores ──")
print(stats_por[:, :6])
```

```python
# Transponer → jugadores × atributos (más legible)
stats_por_T = stats_por.T
print(f"Shape transpuesto: {stats_por_T.shape}  → (jugadores × atributos)")
print("Columnas: [edad, caps, goles]")
print(stats_por_T[:5])
```

---
## 4. Indexación y slicing

En NumPy usamos `[fila, columna]` para acceder a elementos de una matriz 2D.


```python
# Edad del primer jugador (Diogo Costa)
print(f"Edad de {nombres_por[0]}: {edades_por_arr[0]}")

# Últimos 3 jugadores — edad
print(f"Últimas 3 edades Portugal: {edades_por_arr[-3:]}")
```

```python
# Toda la fila 0 de la matriz (edades de todos)
print("Edades de Portugal:")
print(stats_por[0])

# Todos los valores de la columna 0 de la transpuesta (edades de todos)
print("Primeras 5 filas de stats_por_T:")
print(stats_por_T[:5, :])  # filas 0-4, todas las columnas
```

---
## 5. Operaciones vectorizadas

NumPy opera sobre **todo el array a la vez** sin necesidad de bucles `for`.


```python
# Goles por partido (ratio) para Portugal
goles_por_cap_por = np.where(caps_por_arr > 0, goles_por_arr / caps_por_arr, 0).round(3)
print("Goles/cap Portugal:", goles_por_cap_por[:10])
```

```python
# Misma operación para Argentina
goles_por_cap_arg = np.where(caps_arg_arr > 0, goles_arg_arr / caps_arg_arr, 0).round(3)

# ¿Quién tiene el mejor ratio?
idx_max_por = np.argmax(goles_por_cap_por)
idx_max_arg = np.argmax(goles_por_cap_arg)
print(f"Mejor ratio Portugal:  {nombres_por[idx_max_por]} ({goles_por_cap_por[idx_max_por]:.3f} goles/cap)")
print(f"Mejor ratio Argentina: {nombres_arg[idx_max_arg]} ({goles_por_cap_arg[idx_max_arg]:.3f} goles/cap)")
```

---
## 6. Estadísticas descriptivas

| Función | Descripción |
|---|---|
| `np.mean()` | Media aritmética |
| `np.median()` | Mediana |
| `np.std()` | Desviación estándar |
| `np.min()` / `np.max()` | Mínimo / Máximo |
| `np.percentile()` | Percentil |
| `np.sum()` | Suma |


```python
print("=" * 50)
print(f"{'':30} Portugal   Argentina")
print("=" * 50)
print(f"{'Edad promedio':30} {np.mean(edades_por_arr):.1f}       {np.mean(edades_arg_arr):.1f}")
print(f"{'Edad mediana':30} {np.median(edades_por_arr):.1f}       {np.median(edades_arg_arr):.1f}")
print(f"{'Caps promedio':30} {np.mean(caps_por_arr):.1f}      {np.mean(caps_arg_arr):.1f}")
print(f"{'Total goles (selección)':30} {np.sum(goles_por_arr)}         {np.sum(goles_arg_arr)}")
print(f"{'Goles máx (jugador)':30} {np.max(goles_por_arr)}         {np.max(goles_arg_arr)}")
print(f"{'Desv. std edad':30} {np.std(edades_por_arr):.2f}       {np.std(edades_arg_arr):.2f}")
print("=" * 50)
```

```python
# Percentiles de edad — Portugal
p25, p50, p75 = np.percentile(edades_por_arr, [25, 50, 75])
print(f"Portugal — Edad P25: {p25:.0f}  P50: {p50:.0f}  P75: {p75:.0f}")

p25, p50, p75 = np.percentile(edades_arg_arr, [25, 50, 75])
print(f"Argentina — Edad P25: {p25:.0f}  P50: {p50:.0f}  P75: {p75:.0f}")
```

---
## 7. Filtrado booleano (máscaras)

Un array booleano actúa como **máscara** para filtrar elementos.


```python
# Jugadores de Argentina mayores de 30
mascara_veteranos = edades_arg_arr > 30
print("Máscara (True = veterano):", mascara_veteranos[:10])
print()
nombres_arr_arg = np.array(nombres_arg)
veteranos = nombres_arr_arg[mascara_veteranos]
edades_veteranos = edades_arg_arr[mascara_veteranos]
caps_veteranos   = caps_arg_arr[mascara_veteranos]

for n, e, c in zip(veteranos, edades_veteranos, caps_veteranos):
    print(f"  {n:25} {e} años  |  {c} caps")
```

```python
# Goleadores de Portugal (más de 5 goles)
nombres_arr_por = np.array(nombres_por)
mascara_goleadores = goles_por_arr > 5
print("Goleadores de Portugal (>5 goles):")
for n, g in zip(nombres_arr_por[mascara_goleadores], goles_por_arr[mascara_goleadores]):
    print(f"  {n:30} {g} goles")
```

---
## 8. Ordenamiento — `np.argsort()`

`np.argsort()` devuelve los **índices** que ordenarían el array,  
muy útil para reordenar arrays relacionados.


```python
# Top 5 goleadores de Argentina
orden = np.argsort(goles_arg_arr)[::-1]   # descendente
print("Top 5 goleadores de Argentina:")
for i in orden[:5]:
    print(f"  {nombres_arg[i]:25} {goles_arg_arr[i]:3} goles  |  {caps_arg_arr[i]} caps")
```

```python
# Jugador más joven y más veterano — Portugal
idx_menor = np.argmin(edades_por_arr)
idx_mayor = np.argmax(edades_por_arr)
print(f"Más joven:   {nombres_por[idx_menor]:25} ({edades_por_arr[idx_menor]} años)")
print(f"Más veterano:{nombres_por[idx_mayor]:25} ({edades_por_arr[idx_mayor]} años)")
```

---
## 9. Concatenación y apilado de arrays

| Función | Descripción |
|---|---|
| `np.concatenate()` | Une arrays en un mismo eje |
| `np.stack()` | Apila arrays en un nuevo eje |
| `np.vstack()` | Apila en vertical (por filas) |
| `np.hstack()` | Apila en horizontal (por columnas) |


```python
# Unir edades de ambas selecciones
edades_total = np.concatenate([edades_por_arr, edades_arg_arr])
print(f"Total jugadores: {len(edades_total)}")
print(f"Edad promedio global: {np.mean(edades_total):.2f} años")
print(f"Rango de edades: {np.min(edades_total)} – {np.max(edades_total)} años")
```

```python
# Combinar caps y goles en una sola matriz para Portugal
stats_2col = np.column_stack([caps_por_arr, goles_por_arr])
print("Caps | Goles (primeros 8 jugadores):")
print(stats_2col[:8])
```

---
## 10. Álgebra lineal básica con `np.linalg`


```python
# Correlación entre caps y goles — Portugal
corr_por = np.corrcoef(caps_por_arr, goles_por_arr)[0, 1]
corr_arg = np.corrcoef(caps_arg_arr, goles_arg_arr)[0, 1]
print(f"Correlación caps ↔ goles — Portugal:  {corr_por:.4f}")
print(f"Correlación caps ↔ goles — Argentina: {corr_arg:.4f}")
print()
print("(1 = perfecta correlación positiva, 0 = sin correlación)")
```

```python
# Distancia euclidiana entre perfiles promedio de ambas selecciones
perfil_por = np.array([np.mean(edades_por_arr), np.mean(caps_por_arr), np.mean(goles_por_arr)])
perfil_arg = np.array([np.mean(edades_arg_arr), np.mean(caps_arg_arr), np.mean(goles_arg_arr)])

distancia = np.linalg.norm(perfil_por - perfil_arg)
print(f"Perfil Portugal:  edad={perfil_por[0]:.1f}  caps={perfil_por[1]:.1f}  goles={perfil_por[2]:.1f}")
print(f"Perfil Argentina: edad={perfil_arg[0]:.1f}  caps={perfil_arg[1]:.1f}  goles={perfil_arg[2]:.1f}")
print(f"
Distancia euclidiana entre planteles: {distancia:.2f}")
```

## 11. Generación de datos con `np.random`

Útil para simulaciones, pruebas y análisis Monte Carlo.


```python
np.random.seed(2026)  # Reproducibilidad

# Simular 10.000 predicciones de goles para el Mundial
# Basado en la media y desv. estándar real de Portugal
media    = np.mean(goles_por_arr)
std_dev  = np.std(goles_por_arr)
simulaciones = np.random.normal(loc=media, scale=std_dev, size=10_000)
simulaciones = np.clip(simulaciones, 0, None)  # sin valores negativos

print(f"Media real goles/jugador Portugal: {media:.2f}")
print(f"Media simulada (10k sims):         {simulaciones.mean():.2f}")
print(f"Probabilidad de >20 goles/jugador: {(simulaciones > 20).mean()*100:.1f}%")
```
