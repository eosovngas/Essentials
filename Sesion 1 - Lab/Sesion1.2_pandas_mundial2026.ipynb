## 2. Crear un DataFrame

`pd.DataFrame()` convierte la lista de diccionarios en una tabla con filas y columnas.


````python
df_por = pd.DataFrame(jugadores_portugal)
df_arg = pd.DataFrame(jugadores_argentina)

# Agregar columna de selección
df_por["seleccion"] = "Portugal"
df_arg["seleccion"] = "Argentina"

print("── Primeras filas de Portugal ──")
df_por.head()
````

````python
print("── Primeras filas de Argentina ──")
df_arg.head()
````

---
## 3. Inspección básica del DataFrame

| Método | Qué hace |
|---|---|
| `.shape` | Filas × columnas |
| `.dtypes` | Tipo de cada columna |
| `.info()` | Resumen general |
| `.describe()` | Estadísticas numéricas |


````python
print(f"Forma del DataFrame Portugal: {df_por.shape}")
print(f"Forma del DataFrame Argentina: {df_arg.shape}")
print()
print("── Tipos de datos ──")
print(df_por.dtypes)
````

````python
df_por.info()
````

````python
print("── Estadísticas descriptivas – Portugal ──")
df_por.describe().round(2)
````

---
## 4. Selección de columnas y filas

### 4.1 Seleccionar columnas


````python
# Una columna → Serie
print(df_por["nombre"])
````

````python
# Varias columnas → DataFrame
df_por[["nombre", "posicion", "club", "goles"]]
````

### 4.2 Selección por posición — `.iloc[]`

````python
# Fila 0 (Diogo Costa)
print(df_por.iloc[0])
````

````python
# Filas 0-4, todas las columnas
df_por.iloc[0:5]
````

### 4.3 Selección por etiqueta — `.loc[]`

````python
# Establecer 'nombre' como índice
df_por2 = df_por.set_index("nombre")
df_por2.loc["Cristiano Ronaldo"]
````

---
## 5. Filtrado con condiciones booleanas


````python
# Delanteros de Portugal
delanteros_por = df_por[df_por["posicion"] == "Delantero"]
delanteros_por[["nombre", "club", "goles", "caps"]]
````

````python
# Jugadores de Argentina con más de 30 caps
veteranos_arg = df_arg[df_arg["caps"] > 30]
veteranos_arg[["nombre", "posicion", "club", "caps"]].sort_values("caps", ascending=False)
````

````python
# Jugadores de la Premier League (Portugal)
epl_clubs = ["Manchester City", "Manchester United", "Chelsea", "Arsenal", "Liverpool", "Tottenham", "Wolverhampton"]
df_por[df_por["club"].isin(epl_clubs)][["nombre", "club", "posicion"]]
````

---
## 6. Ordenamiento — `.sort_values()`


````python
# Top 5 goleadores de Argentina
df_arg.sort_values("goles", ascending=False)[["nombre", "posicion", "club", "goles"]].head(5)
````

````python
# Ordenar por posición y luego por edad
df_por.sort_values(["posicion", "edad"])[["nombre", "posicion", "edad", "club"]].reset_index(drop=True)
````

---
## 7. Agregar columnas derivadas


````python
# Ratio goles por partido para Argentina
df_arg["goles_por_cap"] = (df_arg["goles"] / df_arg["caps"]).round(3)

# Categoría de edad
def categoria_edad(edad):
    if edad < 24:
        return "Joven"
    elif edad < 30:
        return "Prime"
    else:
        return "Veterano"

df_arg["categoria"] = df_arg["edad"].apply(categoria_edad)
df_por["categoria"] = df_por["edad"].apply(categoria_edad)

df_arg[["nombre", "goles", "caps", "goles_por_cap", "categoria"]].sort_values("goles_por_cap", ascending=False).head(8)
````

---
## 8. Agrupación — `.groupby()`

Permite calcular estadísticas por categoría.


````python
# Promedio de caps por posición — Portugal
df_por.groupby("posicion")[["caps", "goles", "edad"]].mean().round(2)
````

````python
# Número de jugadores por posición — Argentina
df_arg.groupby("posicion")["nombre"].count().rename("cantidad")
````

````python
# Máximo goleador por posición — Argentina
df_arg.groupby("posicion")["goles"].max()
````

---
## 9. Concatenar DataFrames — `pd.concat()`


````python
# Unir ambas selecciones en un solo DataFrame
df_mundial = pd.concat([df_por, df_arg], ignore_index=True)
print(f"Total jugadores en el dataset: {len(df_mundial)}")
df_mundial[["nombre", "seleccion", "posicion", "club", "goles"]].tail(6)
````

````python
# Comparación de promedios por selección
comparacion = df_mundial.groupby("seleccion")[["edad", "caps", "goles"]].mean().round(2)
comparacion.columns = ["Edad promedio", "Caps promedio", "Goles promedio"]
comparacion
````

---
## 10. Manejo de valores nulos — `.isna()` / `.fillna()`


````python
# Verificar si hay nulos en el dataset
print("Nulos por columna:")
print(df_mundial.isna().sum())
````

````python
# Simular un valor nulo y rellenarlo
import numpy as np
df_test = df_mundial[["nombre", "goles"]].copy()
df_test.loc[0, "goles"] = np.nan
print("Antes de fillna:", df_test["goles"].isna().sum(), "nulo(s)")
df_test["goles"] = df_test["goles"].fillna(0)
print("Después de fillna:", df_test["goles"].isna().sum(), "nulo(s)")
````

## Resumen de métodos aprendidos

| Operación | Código |
|---|---|
| Crear DataFrame | `pd.DataFrame(lista_de_dicts)` |
| Ver primeras filas | `df.head(n)` |
| Estadísticas | `df.describe()` |
| Filtrar | `df[df["col"] == valor]` |
| Seleccionar cols | `df[["col1", "col2"]]` |
| Ordenar | `df.sort_values("col")` |
| Nueva columna | `df["nueva"] = expresion` |
| Aplicar función | `df["col"].apply(func)` |
| Agrupar | `df.groupby("col").agg(...)` |
| Concatenar | `pd.concat([df1, df2])` |
| Exportar CSV | `df.to_csv("archivo.csv")` |
