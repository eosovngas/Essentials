## ¿Qué es WordCloud?

**WordCloud** es una librería de Python que genera nubes de palabras: una representación visual de un texto donde cada palabra aparece con un tamaño proporcional a su frecuencia (o importancia) dentro del texto. Mientras más veces se repite una palabra, más grande se muestra.

Se usa principalmente para explorar rápidamente los temas dominantes en textos libres (encuestas, comentarios, respuestas abiertas, reseñas, etc.), sin tener que leer cada respuesta una por una.

**Parámetros más comunes:**
- `width` / `height`: tamaño de la imagen generada.
- `background_color`: color de fondo de la nube.
- `stopwords`: conjunto de palabras a excluir (artículos, preposiciones, conectores, etc.) para que no dominen la visualización.
- `collocations`: si es `False`, evita agrupar pares de palabras que aparecen juntas frecuentemente.
- `max_words`: número máximo de palabras a mostrar.

**Documentación oficial:** https://amueller.github.io/word_cloud/

## ¿Qué es pandas?

**pandas** es una librería de Python para el análisis y manipulación de datos tabulares (filas y columnas), similar a trabajar con una hoja de cálculo pero de forma programática. Su estructura principal es el **DataFrame**, que permite cargar, limpiar, transformar y explorar datos de forma eficiente.

Se usa para leer archivos (CSV, Excel, etc.), filtrar y agrupar información, calcular estadísticas, combinar distintas fuentes de datos y preparar la información antes de analizarla o visualizarla.

**Funciones más comunes:**
- `pd.read_csv()` / `pd.read_excel()`: cargar datos desde un archivo.
- `df.head()`: ver las primeras filas de un DataFrame.
- `df.dropna()`: eliminar valores vacíos o nulos.
- `df["columna"]`: seleccionar una columna específica.
- `df.groupby()`: agrupar datos para calcular resúmenes.
- `str.cat(sep=" ")`: unir el texto de una columna en un solo string.

**Documentación oficial:** https://pandas.pydata.org/docs/

# Paso `4` -  Construir funcion en python para emplear libreria wordcloud

## Función `generar_nube`

Recibe el nombre de una **columna**, un **título** y el nombre del **archivo** de salida, y genera la nube de palabras correspondiente:

1. Une todo el texto de la columna en un solo string (`str.cat`), descartando valores vacíos (`dropna`).
2. Crea el objeto `WordCloud` con el tamaño de imagen, el fondo blanco y la lista de `stopwords` a excluir, y genera la nube a partir del texto (`.generate`).
3. Dibuja la imagen con `matplotlib` (`imshow`), sin ejes (`axis("off")`) y con el título indicado.
4. Guarda la imagen como PNG (`savefig`) y la muestra en pantalla (`show`).
5. Imprime un mensaje confirmando que se guardó el archivo.

## Parámetros de `WordCloud`

```python
wc = WordCloud(width=1200, height=700, background_color="white",
                stopwords=stopwords, collocations=False).generate(texto)
```

- **`width`, `height`**: ancho y alto de la imagen generada (en píxeles). Aquí `1200x700`.
- **`background_color`**: color de fondo de la nube. `"white"` en este caso.
- **`stopwords`**: conjunto de palabras que se excluyen del conteo (artículos, preposiciones, conectores, etc.), definido antes en la variable `stopwords`.
- **`collocations`**: si es `False`, evita que la librería agrupe automáticamente pares de palabras que aparecen juntas frecuentemente (por ejemplo, evitar que "power bi" se muestre como una sola unidad si no se quiere ese comportamiento).
- **`.generate(texto)`**: método que procesa el texto recibido, cuenta la frecuencia de cada palabra (aplicando las `stopwords`) y arma la nube de palabras lista para dibujar.


````python
# --- 4. Función: genera la nube de una columna y la guarda como PNG ---
def generar_nube(columna, titulo, archivo):
    texto = df[columna].dropna().str.cat(sep=" ")
    wc = WordCloud(width=1200, height=700, background_color="white",
                    stopwords=stopwords, collocations=False).generate(texto)
 
    plt.figure(figsize=(12, 7))
    plt.imshow(wc, interpolation="bilinear")
    plt.axis("off")
    plt.title(titulo)
    plt.savefig(archivo, bbox_inches="tight")
    plt.show()   # muestra la nube en pantalla
    print(f"Guardado: {archivo}")
````

# Paso `7` -  Generar funcion para obtener la frecuencia de repetición de las palabras
## Función `generar_nube` (con DataFrame de frecuencias)

Recibe el nombre de una **columna**, un **título** y el nombre del **archivo** de salida. Genera la nube de palabras, la guarda como PNG y además retorna un DataFrame con las palabras y su frecuencia:

1. Une todo el texto de la columna en un solo string (`str.cat`), descartando valores vacíos (`dropna`).
2. Crea el objeto `WordCloud` con el tamaño de imagen, el fondo blanco y la lista de `stopwords`, **sin generar la nube todavía**.
3. Usa `wc.process_text(texto)` para obtener el conteo de cada palabra en el texto (ya filtrando las `stopwords`), guardándolo en el diccionario `frecuencias` (`{palabra: conteo}`).
4. Con `wc.generate_from_frequencies(frecuencias)` arma la nube a partir de ese diccionario, en vez de volver a procesar el texto — esto permite reutilizar el mismo conteo tanto para la imagen como para el DataFrame.
5. Dibuja la imagen con `matplotlib` (`imshow`), sin ejes (`axis("off")`) y con el título indicado.
6. Guarda la imagen como PNG (`savefig`) y la muestra en pantalla (`show`).
7. Imprime un mensaje confirmando que se guardó el archivo.
8. Convierte el diccionario `frecuencias` en un DataFrame con columnas `palabra` y `frecuencia`, lo ordena de mayor a menor (`sort_values`) y reinicia el índice (`reset_index`).
9. Retorna ese DataFrame, para poder revisarlo, exportarlo o cruzarlo con otros datos.


````python
# --- 5. Función: genera la nube de una columna, la guarda como PNG, devuelve el conjunto de palabras (con su frecuencia) en un DataFrame ---
def generar_nube(columna, titulo, archivo):
    texto = df[columna].dropna().str.cat(sep=" ")
    wc = WordCloud(width=1200, height=700, background_color="white",
                    stopwords=stopwords, collocations=False)
 
    frecuencias = wc.process_text(texto)   # dict {palabra: conteo}
    wc.generate_from_frequencies(frecuencias)
 
    plt.figure(figsize=(12, 7))
    plt.imshow(wc, interpolation="bilinear")
    plt.axis("off")
    plt.title(titulo)
    plt.savefig(archivo, bbox_inches="tight")
    plt.show()   # muestra la nube en pantalla
    print(f"Guardado: {archivo}")
 
    df_palabras = (
        pd.DataFrame(frecuencias.items(), columns=["palabra", "frecuencia"])
        .sort_values("frecuencia", ascending=False)
        .reset_index(drop=True)
    )
    return df_palabras
````
